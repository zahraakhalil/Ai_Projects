```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
df=pd.read_csv('creditcard.csv')
df.head()
```

```python
print("data shape:")
df.shape
```

```python
print("data info:")
df.info()
```

```python
print("data description:")
df.describe()
```

```python
print("missing values in each column:")
df.isnull().sum()
```

```python
print("number of duplicate rows:")
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
print("new data shape after removing duplicates:")
df.shape
```

```python
sns.histplot(df['Amount'], bins=50)
plt.title('Transaction Amount Distribution')
plt.show()
sns.histplot(df['Time'], bins=50)
plt.title('Transaction Time Distribution')
plt.show()
```

```python
print("class value counts:")
df['Class'].value_counts()
```

```python
from sklearn.preprocessing import MinMaxScaler
X=df.drop('Class', axis=1)
mms=MinMaxScaler()
X_scaled= pd.DataFrame(mms.fit_transform(X), columns=X.columns)

X_scaled.head()
```

```python
from sklearn.decomposition import PCA

# Fit PCA
pca = PCA()
pca_df = pd.DataFrame(pca.fit_transform(X_scaled))

# Plot cumulative explained variance
plt.step(range(1, len(pca.explained_variance_ratio_) + 1), np.cumsum(pca.explained_variance_ratio_), where='mid')
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance Ratio')
plt.title('Variance Covered by Principal Components')
plt.grid(True)
plt.show()
```

```python
optimal_components = np.argmax(np.cumsum(pca.explained_variance_ratio_) >= 0.95) + 1
print(f"Optimal number of components: {optimal_components}")
```

```python
pca_df = pca_df.drop(pca_df.columns[14:31], axis=1)
pca_df.head()
```

```python
from sklearn.ensemble import IsolationForest
X = pca_df.sample(n=100000, random_state=42)
iso_forest = IsolationForest(contamination=0.05, random_state=42)
X['Anomaly'] = iso_forest.fit_predict(X)

# Anomalies are labeled as -1
anomalies = X[X['Anomaly'] == -1]

print(f"Number of anomalies detected: {len(anomalies)}")
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
scatter = ax.scatter(X.index, X[0], c=X['Anomaly'], cmap='coolwarm')
plt.colorbar(scatter, ax=ax, label='Anomaly Score')

ax.set_title('Anomaly Detection using Isolation Forest')
ax.set_xlabel('Index')
ax.set_ylabel('Principal Component 1')
plt.show()
```

```python
from sklearn.svm import OneClassSVM
sample_df = pca_df.sample(n=100000, random_state=42)

# Train OneClassSVM on the sampled data
svm = OneClassSVM(kernel='rbf', gamma='auto', nu=0.05)
sample_df['Anomaly'] = svm.fit_predict(sample_df)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
scatter = ax.scatter(sample_df.index, sample_df[0], c=sample_df['Anomaly'], cmap='coolwarm')
plt.colorbar(scatter, ax=ax, label='Anomaly Score')

ax.set_title('Anomaly Detection using One-Class SVM')
ax.set_xlabel('Index')
ax.set_ylabel('Principal Component 1')
plt.show()
```
