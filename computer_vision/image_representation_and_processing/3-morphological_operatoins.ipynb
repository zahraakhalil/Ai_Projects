```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
```

```python
def load_image():
    blank_image = np.zeros((600,600))
    cv.putText(blank_image,text="ABCDE", org = (50,300), fontFace=cv.FONT_HERSHEY_SIMPLEX, fontScale=5,
               color = (255, 255, 255), thickness=25, lineType=cv.LINE_AA)
    return blank_image
```

```python
def display_image(img):
    fig = plt.figure(figsize=(12,10))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap='gray')
```

```python
img = load_image()
display_image(img)
```

### Erosion

```python
kernel = np.ones((5,5), np.uint8)
erosion1 = cv.erode(img, kernel, iterations=1)
display_image(erosion1)
```

### Dilation


```python
kernel = np.ones((5,5), np.uint8)
dilation1 = cv.dilate(img, kernel, iterations=1)
display_image(dilation1)
```

### Openning

```python
# openning = erosion + dilation
img = load_image()
white_noise = np.random.randint(low=0,high=2,size=(600,600))
white_noise 
```

```python
white_noise= white_noise*255
display_image(white_noise)
```

```python
noise_image = white_noise + img
display_image(noise_image)
```

```python
openning = cv.morphologyEx(noise_image, cv.MORPH_OPEN, kernel)
display_image(openning)
```

```python
img = load_image()
black_noise = np.random.randint(low=0, high=2, size=(600,600))
black_noise = black_noise*-255
```

```python
black_noise_img = black_noise + img
display_image(black_noise_img)
```

```python
black_noise_img[black_noise_img==-255] = 0
display_image(black_noise_img)
```

```python
closing = cv.morphologyEx(black_noise_img, cv.MORPH_CLOSE, kernel)
display_image(closing)
```

```python
img = load_image()
display_image(img)
```

```python
gradient = cv.morphologyEx(img, cv.MORPH_GRADIENT, kernel)
display_image(gradient)
```
