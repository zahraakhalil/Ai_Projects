### Drawing

```python
canvas = np.zeros((300, 300, 3), dtype="uint8")
```

```python
green = (0, 255, 0)
cv.line(canvas, (0,0), (300,300), green, 1)
cv.imshow("canvas", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
red = (0, 0, 255)
cv.line(canvas, (0,300), (300,0), red, 1)
cv.imshow("canvas", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
blue = (255, 0 ,0)
cv.rectangle(canvas, (120,195), (180,105), blue, 5)
cv.imshow("canvas", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
blue = (85, 0 ,0)
cv.rectangle(canvas, (120,195), (180, 105),blue, -1)
cv.imshow("canvas", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
canvas = np.zeros((300, 300, 3), dtype="uint8")
(centerX, centerY) = (canvas.shape[1]//2, canvas.shape[0]//2)
white = (255, 255, 255)
for r in range(0, 150, 25):
    cv.circle(canvas, (centerX, centerY), r, white, 1)
cv.imshow("canvad", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```
